## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

* **Rule-based approach limitations:** Simple rules (e.g., `if word_count > 1000 and keyword_density > 2%`) fail because search intent is complex and semantic. High word count alone doesn't guarantee value, and overusing keywords leads to spam penalties.
* **ML Value:** A machine learning model learns non-linear interactions between text quality, semantic relevance, structure, and user engagement proxies that static `if/else` conditions cannot capture.
